```python
import pandas as pd

# Load raw digitised data
df = pd.read_csv("../data/processed/bubble_curves.csv")
df.head()
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt

# Replicate Figure 10 of the paper, using different markers for each condition
styles = {
    "f1": {"marker": "s"},
    "f0_98": {"marker": "d"},
    "f0_93": {"marker": "x"},
    "f0_84": {"marker": "^"},
}

# Loop through each condition
for condition, group in df.groupby("condition_id"):
    
    style = styles.get(condition, {})  # fallback if missing
    
    plt.scatter(
        group["time_s"],
        group["radius_um"],
        **style
    )

# Labels
plt.xlabel("Time (s)")
plt.ylabel("Radius (µm)")
plt.title("Bubble Dissolution Curves (Figure 10)")
plt.grid(True)
plt.show()
```
