```python
import pandas as pd

# Load raw digitised data
df = pd.read_csv("../data/processed/bubble_curves.csv")
df.head()
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt

# Replicate Figure 7 of the paper, using different markers for each condition
styles = {
    "fig7_cross": {"marker": "x"},
    "fig7_square": {"marker": "s"},
    "fig7_circle": {"marker": "o"},
}

# Loop through each condition
for condition, group in df.groupby("condition_id"):
    
    style = styles.get(condition, {})  # fallback if missing
    
    plt.scatter(
        group["time_s"],
        group["radius_um"],
        **style
    )

# Labels
plt.xlabel("Time (s)")
plt.ylabel("Radius (µm)")
plt.title("Bubble Dissolution Curves (Figure 7)")
plt.legend()
plt.grid(True)
plt.show()
```
